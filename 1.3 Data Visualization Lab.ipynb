```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('spotify-2023.csv', encoding='ISO-8859-1')

df.head()
```

```python
df['bpm'].plot(kind='hist', bins=20)
plt.title('Distribution of BPM')
plt.xlabel('BPM')
plt.ylabel('Frequency')
plt.show()
```

```python
df['key'].value_counts().sort_index().plot(kind='bar')
plt.title('Distribution of Key')
plt.xlabel('Key')
plt.ylabel('Count')
plt.show()
```

```python
df['mode'].value_counts().plot(kind='bar')
plt.title('Distribution of Mode (0 = Minor, 1 = Major)')
plt.xlabel('Mode')
plt.ylabel('Count')
plt.show()
```

```python
hitfilter = df[
    (df['bpm'].between(120, 130)) &
    (df['key'] == 'C#')
]

hitfilter[['track_name', 'artist(s)_name', 'bpm', 'key']]
```
